```python
# DATA CLEANING

# Dropping unwanted column
if 'Unnamed: 0' in df.columns:
    df = df.drop(['Unnamed: 0','id'], axis=1)
df
```

```python
#null values
print(df.isnull().sum())
```

```python
#NULL VALUES HANDLING
df['Arrival Delay in Minutes']=df['Arrival Delay in Minutes'].fillna(df['Arrival Delay in Minutes'].median())
```

```python
#checking for duplicate columns
print(df.duplicated().sum())
```

```python
#OUTLIERS

#A. checking outliers using boxplot
import matplotlib.pyplot as plt
plt.figure(figsize=(15,10))
df.select_dtypes(exclude=object).boxplot()
plt.xticks(rotation=45)
plt.title("box plot showing outliers")
plt.show()
```

```python
#b. number of outliers in each column
col=df.select_dtypes(exclude=object).columns
for i in col:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outliers=df[(df[i]<lower)| (df[i]>upper)]
  print(i,"outliers:",len(outliers))
```

```python
#c. treating the outliers
for i in col:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  df[i]=df[i].clip(lower,upper)
```

```python
#d. boxplot after treating outliers
plt.figure(figsize=(15,10))
df.select_dtypes(exclude=object).boxplot()
plt.xticks(rotation=45)
plt.title("box plot after outliers treatments")
plt.show()
```

```python
#LABEL ENCODING
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
obj=df.select_dtypes(include=object).columns
for i in obj:
  df[i]=le.fit_transform(df[i])
df
```

```python
#assigning input feature
x=df
x
```

```python
#feature scaling
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
#KMEANS MODEL BUILDING
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters=i,init='k-means++',random_state=41)
  #training
  kmeans.fit(x_scaled)
  #inertia
  wcss.append(kmeans.inertia_)
wcss
```

```python
#ELBOW METHOD
import matplotlib.pyplot as plt
plt.plot(range(1,11),wcss)
plt.xlabel('K- Value')
plt.ylabel("value of inertia ")
plt.show()
```

```python
#model building using the obtained k value
kmeans=KMeans(n_clusters=7,init='k-means++',random_state=41)
#training
clust_model=kmeans.fit(x_scaled)
labels=clust_model.labels_
print(labels)
df['cluster']=kmeans.labels_
df
```

```python
print(df['cluster'].value_counts())
```

```python
#PERFORMANCE EVALUATION
from sklearn.metrics import silhouette_samples
print(silhouette_samples(x_scaled,labels))
```
