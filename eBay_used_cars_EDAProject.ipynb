```python
#DATA CLEANING

#1.removing duplicate row
print(df.duplicated().sum())
df=df.drop_duplicates()
```

```python
#2.DROPPING USELESS COLUMNS (same value in almost every row)
print(df["seller"].value_counts())
print(df["offerType"].value_counts())
print(df["nrOfPictures"].value_counts())
df = df.drop(["seller", "offerType", "nrOfPictures"],axis=1)
df
```

```python
#3.FIXING 'PRICE' COLUMN
df["price"] = df["price"].astype(str)
df["price"] = df["price"].str.replace("$", "", regex=False)
df["price"] = df["price"].str.replace(",", "", regex=False)
df["price"] = pd.to_numeric(df["price"])
print(df['price'].dtype)
```

```python
#4.FIXING 'ODOMETER' COLUMN
df["odometer"] = df["odometer"].astype(str)
df["odometer"] = df["odometer"].str.replace("km", "", regex=False)
df["odometer"] = df["odometer"].str.replace(",", "", regex=False)
df["odometer"] = pd.to_numeric(df["odometer"])
print(df["odometer"].dtype)
```

```python
#5. FIX DATE COLUMNS (text -> real dates)
df["dateCrawled"] = pd.to_datetime(df["dateCrawled"])
df["dateCreated"] = pd.to_datetime(df["dateCreated"])
df["lastSeen"] = pd.to_datetime(df["lastSeen"])
```

```python
#6.FIX WRONG VALUES OF PRICE
print("Rows before price filter:", len(df))
df = df[(df["price"] >= 100) & (df["price"] <= 100000)]
print("Rows after price filter:", len(df))
```

```python
#7.FIX WRONG VALUES OF OTHER COLUMNS
df.loc[(df["yearOfRegistration"] < 1950) | (df["yearOfRegistration"] > 2016), "yearOfRegistration"] = np.nan
df.loc[(df["powerPS"] < 10) | (df["powerPS"] > 1000), "powerPS"] = np.nan
df.loc[(df["monthOfRegistration"] < 1) | (df["monthOfRegistration"] > 12), "monthOfRegistration"] = np.nan
```

```python
#8. HANDLE MISSING VALUES
print(df.isnull().sum())

#A. Text columns: fill with the most common value (mode)
df["vehicleType"] = df["vehicleType"].fillna(df["vehicleType"].mode()[0])
df["gearbox"] = df["gearbox"].fillna(df["gearbox"].mode()[0])
df["fuelType"] = df["fuelType"].fillna(df["fuelType"].mode()[0])
```

```python
#B. Text columns where guessing can create error-null values converted into unknown
df["model"] = df["model"].fillna("unknown")
df["notRepairedDamage"] = df["notRepairedDamage"].fillna("unknown")
```

```python
#C. Number columns: fill with the middle value (median)
df["powerPS"] = df["powerPS"].fillna(df["powerPS"].median())
df["yearOfRegistration"] = df["yearOfRegistration"].fillna(df["yearOfRegistration"].median())
df["monthOfRegistration"] = df["monthOfRegistration"].fillna(df["monthOfRegistration"].mode()[0])
print(df.isnull().sum())
```

```python
# 9. CREATING NEW COLUMNS
df["car_age"] = 2016 - df["yearOfRegistration"]
df["price_per_km"] = df["price"] / df["odometer"]
df["weekday"] = df["dateCreated"].dt.day_name()
print(df.shape)
print(df.head())
```

```python
#DATA VISUALIZATION

#1. Price distribution
plt.figure(figsize=(8, 5))
sns.histplot(df["price"], bins=50, kde=True)
plt.title("Price distribution")
plt.show()
```

```python
#2. Top 10 brands
plt.figure(figsize=(8, 5))
df["brand"].value_counts().head(10).plot(kind="barh")
plt.title("Top 10 brands")
plt.xlabel("Number of cars")
plt.show()
```

```python
#3. Fuel type share
plt.figure(figsize=(6, 6))
df["fuelType"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Fuel type")
plt.ylabel("")
plt.show()
```

```python
#4. Gearbox count
plt.figure(figsize=(6, 4))
sns.countplot(x="gearbox", data=df)
plt.title("Gearbox type")
plt.show()
```

```python
#5. Average price of the top 10 brands
top_brands = df["brand"].value_counts().head(10).index
top_df = df[df["brand"].isin(top_brands)]

plt.figure(figsize=(8, 5))
top_df.groupby("brand")["price"].mean().sort_values().plot(kind="barh")
plt.title("Average price by brand")
plt.xlabel("Average price")
plt.show()
```

```python
#6. Price by vehicle type (box plot)
plt.figure(figsize=(10, 5))
sns.boxplot(x="vehicleType", y="price", data=df)
plt.title("Price by vehicle type")
plt.xticks(rotation=45)
plt.show()
```

```python
#7. Price vs car age (scatter plot)
plt.figure(figsize=(8, 5))
sns.scatterplot(x="car_age", y="price", data=df.sample(3000, random_state=1), alpha=0.4)
plt.title("Price vs car age")
plt.show()
```

```python
#8. Price vs odometer
plt.figure(figsize=(8, 5))
sns.boxplot(x="odometer", y="price", data=df, showfliers=False)
plt.title("Price vs kilometers driven")
plt.xticks(rotation=45)
plt.show()
```

```python
#9. Ads per weekday
plt.figure(figsize=(8, 4))
sns.countplot(x="weekday", data=df)
plt.title("Ads created per weekday")
plt.xticks(rotation=45)
plt.show()
```

```python
#10. Correlation heatmap
number_columns = ["price", "yearOfRegistration", "powerPS", "odometer", "car_age"]
plt.figure(figsize=(8, 6))
sns.heatmap(df[number_columns].corr(), annot=True, cmap="coolwarm")
plt.title("Correlation between number columns")
plt.show()
```

```python
#11.Vehicle type vs fuel type (count heatmap)
table = pd.crosstab(df["vehicleType"], df["fuelType"])
plt.figure(figsize=(9, 6))
sns.heatmap(table, annot=True, fmt="d", cmap="Blues")
plt.title("Vehicle type vs fuel type")
plt.show()
```

```python
#12. Power vs price
plt.figure(figsize=(8, 5))
sns.scatterplot(x="powerPS", y="price", data=df.sample(3000, random_state=1), alpha=0.4)
plt.title("Price vs engine power")
plt.show()
```

```python
#13.Does unrepaired damage lower the price?
plt.figure(figsize=(6, 4))
df.groupby("notRepairedDamage")["price"].mean().plot(kind="bar")
plt.title("Average price by unrepaired damage")
plt.ylabel("Average price")
plt.show()
```

```python
# SIMPLE INSIGHTS
print("Average price:", round(df["price"].mean()))
print("Most expensive brand (average):", top_df.groupby("brand")["price"].mean().idxmax())
print("Average price - manual (manuell):", round(df[df["gearbox"] == "manuell"]["price"].mean()))
print("Average price - automatic (automatik):", round(df[df["gearbox"] == "automatik"]["price"].mean()))
```
