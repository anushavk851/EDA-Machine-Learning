```python
#DATA CLEANING

#removing unwanted column
df=df.drop('CUST_ID',axis=1)
df
```

```python
#min value of each columns
for i in df.columns:
  print(i,":",df[i].min())
```

```python
#NULL VALUES HANDLING
for i in df.columns:
  df[i]=df[i].fillna(df[i].median())
print(df.isnull().sum())
```

```python
#OUTLIERS

#A. Boxplot to detect outliers
plt.figure(figsize=(10,8))
df.boxplot()
plt.xticks(rotation=45)
plt.title("BoxPlot before outliers treatment")
plt.show()
```

```python
#B. number of outliers in each column
for i in df.columns:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outliers=df[(df[i]<lower)|(df[i]>upper)]
  print(i,"outliers:",len(outliers))
```

```python
#C. treating outliers using clip function
for i in df.columns:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  df[i]=df[i].clip(lower,upper)
```

```python
#D. Boxplot after treating outliers
plt.figure(figsize=(10,8))
df.boxplot()
plt.xticks(rotation=45)
plt.title("BoxPlot after outliers treatment")
plt.show()
```

```python
#input selection
x=df
```

```python
#FEATURE SCALING
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
#CALCULATING INERTIA OF CURRESPONDING K VALUES
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters=i,init='k-means++',random_state=41)
  kmeans.fit(x_scaled)
  wcss.append(kmeans.inertia_)
wcss
```

```python
#ELBOW METHOD
plt.plot(range(1,11),wcss)
plt.xlabel("K Values")
plt.ylabel("Inertia")
plt.show()
```

```python
#MODEL BUILDING WHERE K=7
kmeans=KMeans(n_clusters=7,init="k-means++",random_state=41)
clust=kmeans.fit(x_scaled)
labels=clust.labels_
df['cluster']=kmeans.labels_
df
```

```python
#PERFORMANCE EVALUATION
from sklearn.metrics import silhouette_samples
print(silhouette_samples(x_scaled,labels))
```
