```python
#OUTLIERS

#A. boxplot to check outliers
import matplotlib.pyplot as plt
plt.figure(figsize=(15,10))
df.select_dtypes(exclude=object).boxplot()
plt.xticks(rotation=45)
plt.title("Boxplot before treating")
plt.show()
```

```python
#B. check number of outliers
obj=df.select_dtypes(exclude=object).columns
for i in obj:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outliers=df[(df[i]<lower)|(df[i]>upper)]
  print(i,"outliers :",len(outliers))
```

```python
#C. treating outliers using cliping method
for i in obj:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  df[i]=df[i].clip(lower,upper)
```

```python
#D. boxplot to check outliers
import matplotlib.pyplot as plt
plt.figure(figsize=(15,10))
df.select_dtypes(exclude=object).boxplot()
plt.xticks(rotation=45)
plt.title("Boxplot after treating")
plt.show()
```

```python
#Frequency ENCODING
freq = df['city_name'].value_counts()
df['city_name'] = df['city_name'].map(freq)
freq2= df['country'].value_counts()
df['country'] = df['country'].map(freq2)

df
```

```python
#input selection
x=df
x
```

```python
#feature scaling
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
#KMEANS MODEL BUILDING USING RANGE OF K VALUES
from sklearn.cluster import KMeans
wcss=[]
for i in range(2,11):
  kmeans=KMeans(n_clusters=i,init='k-means++',random_state=41)
  #training
  kmeans.fit(x_scaled)
  #inertia
  wcss.append(kmeans.inertia_)
wcss
```

```python
#ELBOW Method
import matplotlib.pyplot as plt
plt.plot(range(2,11),wcss)
plt.xlabel("K-values")
plt.ylabel("inertia value")
plt.show()
```

```python
#MODEL BUILDING OF KMEANS WHERE K=8
kmeans=KMeans(n_clusters=8,init="k-means++",random_state=41)
clust=kmeans.fit(x_scaled)
labels=clust.labels_
print(labels)
df['cluster']=kmeans.labels_
df
```

```python
df['cluster'].value_counts()
```

```python
#PERFORMANCE EVALUATION
from sklearn.metrics import silhouette_samples
print(silhouette_samples(x_scaled,labels))
```
