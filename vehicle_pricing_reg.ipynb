```python
import pandas as pd
import numpy as np
df0=pd.read_csv('/content/Vehicle_Pricing.csv')
df0
```

```python
df=df0.drop(['Unnamed: 0','Location'],axis=1)
df
```

```python
#column names
print(df.columns)
#numerical column names
print(df.select_dtypes(exclude=object).columns)
#non-numerical column names
print(df.select_dtypes(include=object).columns)
```

```python
#target column
print(df['Price'])
```

```python
#DATA CLEANING

#1 Duplicate Values
print(df.duplicated().sum())
df=df.drop_duplicates()
print(df.duplicated().sum())
```

```python
#Dropping new_price column because it has substantial missing values and mixed formats.
df=df.drop('New_Price',axis=1)
df
```

```python
#Mileage unit correction

#1.splitting mileage into number and unit
df['Mileage_num']=df['Mileage'].str.split(' ').str[0].astype(float)
df['Mileage_unit']=df['Mileage'].str.split(' ').str[1]
df['Mileage_unit'].value_counts()
```

```python
#2.converting km/kg to kmpl
df['Mileage']=df['Mileage_num']
#CNG: /1.5 and LPG: /1.43
df.loc[(df['Mileage_unit']=='km/kg') & (df['Fuel_Type']=='CNG'),'Mileage']=df['Mileage_num']/1.5
df.loc[(df['Mileage_unit']=='km/kg') & (df['Fuel_Type']=='LPG'),'Mileage']=df['Mileage_num']/1.43
df=df.drop(['Mileage_unit','Mileage_num'],axis=1)
df
```

```python
#YEAR COLUMN
print(df['Year'].value_counts())
```

```python
#FUEL TYPE COLUMN
print(df['Fuel_Type'].value_counts())
#Transmission column'
print(df['Transmission'].value_counts())
```

```python
#Owner_Type column
print(df['Owner_Type'].value_counts())
```

```python
#ENGINE COLUMN CONVERSION
df['Engine_unit']=df['Engine'].str.split(' ').str[1]
df['Engine']=df['Engine'].str.split(' ').str[0].astype(float)
df['Engine_unit'].value_counts()
df=df.drop('Engine_unit',axis=1)
df
```

```python
#POWER COLUMN
df['Power_unit']=df['Power'].str.split(' ').str[1]
print(df['Power_unit'].value_counts())
df=df.drop('Power_unit',axis=1)
df['Power']=pd.to_numeric(df['Power'].str.split(' ').str[0],errors='coerce')   #converting null to NaN value
df['Power'].isnull().sum()
```

```python
#CONVERTING 0 TO NaN

print(df['Mileage'].min()) #0
df['Mileage']=df['Mileage'].replace(0,np.nan)
print(df['Mileage'].min())

print(df['Engine'].min())
print(df['Power'].min())

print(df['Seats'].min()) #0
df['Seats']=df['Seats'].replace(0,np.nan)
print(df['Seats'].min())
```

```python
#OUTLIERS

#A.Boxplot for numerical columns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 8))
df.drop(['Price','Seats','Year'], axis=1).boxplot()
plt.xticks(rotation=45)
plt.title("Box Plot of Numerical Features")
plt.show()
```

```python
#B.Find the number of outliers
num= df.drop(['Price','Seats','Year'], axis=1).select_dtypes(include='number').columns
for i in num:
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[i] < lower) | (df[i] > upper)]
    print(i, "outliers:", len(outliers))
```

```python
#C. treating outliers
for i in num:
    df[i] = df[i].clip(lower, upper)
```

```python
#D.Boxplot for numerical columns
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 8))
df.drop(['Price','Seats','Year'], axis=1).boxplot()
plt.xticks(rotation=45)
plt.title("Box Plot of Numerical Features")
plt.show()
```

```python
#NULL VALUE HANDLING
print(df.isnull().sum())
null_values=['Mileage','Engine','Power']
for nulls in null_values:
  df[nulls]=df[nulls].fillna(df[nulls].median())
print(df.isnull().sum())
```

```python
#null value handling of 'SEATS'
df['Seats']=df['Seats'].fillna(df['Seats'].mode()[0])
df['Seats']=df['Seats'].astype(int)
print(df['Seats'].dtype)
print(df['Seats'].value_counts())
print(df.isnull().sum())
```

```python
#datatypes after data cleaning
print(df.dtypes)
```

```python
#ENCODING

# Extract Brand
df['Brand'] = df['Name'].str.split().str[0]
# Remove original Name
df.drop('Name', axis=1, inplace=True)
# Convert categorical columns into dummy variables
df = pd.get_dummies(df, drop_first=True)
print(df.shape)
print(df.head())
print(df.dtypes)
```

```python
#FEATURE SELECTION
correlation=df.corr(numeric_only=True)
print(correlation['Price'].sort_values(ascending=False))
```

```python
#INPUT OUTPUT SPLITTING
x=df.drop('Price',axis=1)
y=df['Price']
print(x.ndim)
print(y.ndim)
x
```

```python
#TRAIN TEST SPLITTING
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.30,random_state=41)
print(xtrain.shape)
print(ytrain.shape)
print(xtest.shape)
print(ytest.shape)
```

```python
#LINEAR REGRESSION MODEL BUILDING
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
#model training
lr.fit(xtrain,ytrain)
#model testing
ypred1=lr.predict(xtest)
ypred1
```

```python
ytest.values
```

```python
#REGRESSION PLOT
import matplotlib.pyplot as plt
plt.scatter(ytest,ypred1)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted Price")
plt.show()
```

```python
#DECISION TREE MODEL BUILDING
from sklearn.tree import DecisionTreeRegressor
dt=DecisionTreeRegressor()
#model training
dt.fit(xtrain,ytrain)
#model testing
ypred2=dt.predict(xtest)
ypred2
```

```python
#ENSEMBLE MODEL BUILDING

#1.RANDOM FOREST
from sklearn.ensemble import RandomForestRegressor
rfr=RandomForestRegressor(n_estimators=100,criterion='squared_error',max_depth=10,random_state=41)
#model training
rfr.fit(xtrain,ytrain)
#model testing
ypred3=rfr.predict(xtest)
ypred3
```

```python
#2.XGBoost
from xgboost import XGBRegressor
xgb=XGBRegressor()
#model training
xgb.fit(xtrain,ytrain)
#model testing
ypred4=xgb.predict(xtest)
ypred4
```

```python
ytest.values
```

```python
#FEATURE SCALING

from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(xtrain)
xtrain=scaler.transform(xtrain)
xtrain
```

```python
xtest=scaler.transform(xtest)
xtest
```

```python
#KNN MODEL BUILDING
from sklearn.neighbors import KNeighborsRegressor
knn=KNeighborsRegressor(n_neighbors=10)
#model training
knn.fit(xtrain,ytrain)
#model testing
ypred5=knn.predict(xtest)
ypred5
```

```python
#PERFORMANCE EVALUATION
from sklearn.metrics import mean_squared_error,root_mean_squared_error,mean_absolute_error,mean_absolute_percentage_error,r2_score

#LINEAR REGRESSION
print("MSE:",mean_squared_error(ytest,ypred1))
print("RMSE:",root_mean_squared_error(ytest,ypred1))
print("MAE:",mean_absolute_error(ytest,ypred1))
print("MAPE:",mean_absolute_percentage_error(ytest,ypred1))
print("R2 squared:",r2_score(ytest,ypred1))
```
