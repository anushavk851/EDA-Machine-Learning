```python
#DATA CLEANING

#min values in each numerical columns
num=df.select_dtypes(exclude=object).columns
for i in num:
  print(i,":",df[i].min())
```

```python
#DATE COLUMN
df['Date']=pd.to_datetime(df['Date'],format='mixed')
df['Year']=df['Date'].dt.year
df['Month']=df['Date'].dt.month
df=df.drop('Date',axis=1)
df
```

```python
#null value handling
print(df.isnull().sum())
obj=df.select_dtypes(include=object).columns
num=df.select_dtypes(exclude=object).columns
for i in num:
  df[i]=df[i].fillna(df[i].median())
for ch in obj:
  df[ch]=df[ch].fillna(df[ch].mode()[0])
print(df.isnull().sum())
```

```python
#ONE HOT ENCODING
print(df.select_dtypes(include=object).columns)
df=pd.get_dummies(df,columns=['Ship_Type', 'Route_Type', 'Engine_Type', 'Maintenance_Status','Weather_Condition'],
                  drop_first=True)
print(df.shape)
df
```

```python
#OUTLIERS

#A. VISUALIZATION OF OUTLIERS USING BOXPLOT
plt.figure(figsize=(15,10))
df.select_dtypes(exclude=object).boxplot()
plt.xticks(rotation=45)
plt.title("boxplot before outliers treatment")
```

```python
#B. NUMBER OF OUTLIERS
num=df.select_dtypes(include='number').columns
for i in num:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outliers=df[(df[i]<lower)|(df[i]>upper)]
  print(i,"outliers:",len(outliers))
```

```python
#C. TREATING OUTLIERS
for i in num:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  df[i]=df[i].clip(lower,upper)
```

```python
#D. BOXPLOT AFTER OUTLIERS TREATMENT
plt.figure(figsize=(15,10))
df.select_dtypes(exclude=object).boxplot()
plt.xticks(rotation=45)
plt.title("boxplot after outliers treatment")
```

```python
#INPUT SELECTION
x=df
x
```

```python
#FEATURE SCALING
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
#MODEL BUILDING USING RANGE OF K-VALUES
from sklearn.cluster import KMeans
wcss=[]
for k in range(1,11):
  kmeans=KMeans(n_clusters=k,init="k-means++",random_state=41)
  kmeans.fit(x_scaled)
  wcss.append(kmeans.inertia_)
wcss
```

```python
#ELBOW METHOD
plt.plot(range(1,11),wcss)
plt.xlabel("k-values")
plt.ylabel("inertia")
plt.show()
```

```python
#MODEL BUILDING
kmeans=KMeans(n_clusters=8,init="k-means++",random_state=41)
clust=kmeans.fit(x_scaled)
labels=clust.labels_
print(labels)
df['cluster']=kmeans.labels_
df
```

```python
#PERFORMANCE EVALUATION
from sklearn.metrics import silhouette_samples
print(silhouette_samples(x_scaled,labels))
```
