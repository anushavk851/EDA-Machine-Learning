```python
#DATA CLEANING
#OUTLIERS

#A. BOXPLOT TO CHECK OUTLIERS
plt.figure(figsize=(10,8))
df.boxplot()
plt.xticks(rotation=45)
plt.title("boxplot before outliers treatment")
```

```python
#B. NUMBER OF OUTLIERS
for i in df.columns:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outliers=df[(df[i]<lower)|(df[i]>upper)]
  print(i,"outliers:",len(outliers))
```

```python
#C. TREATING OUTLIERS
for i in df.columns:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  df[i]=df[i].clip(lower,upper)
```

```python
#D. BOXPLOT AFTER OUTLIERS TREATMENT
plt.figure(figsize=(10,8))
df.boxplot()
plt.xticks(rotation=45)
plt.title("boxplot after outliers treatment")
```

```python
#INPUT SELECTION
x=df
x
```

```python
#FEATURE SCALING
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
#FINDING INERTIA OF CORRESPONDING K-VALUE
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,15):
  kmeans=KMeans(n_clusters=i,init="k-means++",random_state=41)
  kmeans.fit(x_scaled)
  wcss.append(kmeans.inertia_)
wcss
```

```python
#ELBOW METHOD
plt.plot(range(1,15),wcss)
plt.xlabel("K-values")
plt.ylabel("inertia")
plt.show()
```

```python
#MODEL BUILDING USING K=9
kmeans=KMeans(n_clusters=9,init="k-means++",random_state=41)
clust=kmeans.fit(x_scaled)
labels=clust.labels_
print(labels)
df['cluster']=kmeans.labels_
df
```

```python
#PERFORMANCE EVALUATION
from sklearn.metrics import silhouette_samples
print(silhouette_samples(x_scaled,labels))
```
