```python
#null values handling
print(df.isnull().sum())
df['Income']=df['Income'].fillna(df['Income'].median())
print(df.isnull().sum())
```

```python
#converting date-time column to month and year
df['Dt_Customer']=pd.to_datetime(df['Dt_Customer'],format='mixed')
df['year']=df['Dt_Customer'].dt.year
df['month']=df['Dt_Customer'].dt.month
df=df.drop('Dt_Customer',axis=1)
df
```

```python
#OUTLIERS

#A. BOXPLOT BEFORE OUTLIERS TREATMENT
plt.figure(figsize=(15,10))
df.select_dtypes(exclude=object).boxplot()
plt.xticks(rotation=45)
plt.title("Boxplot before outliers treatment")
plt.show()
```

```python
#B. NUMBER OF OUTLIERS IN EACH COLUMNS
num=df.select_dtypes(exclude=object).columns
for i in num:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outliers=df[(df[i]<lower)|(df[i]>upper)]
  print(i,"outliers:",len(outliers))
```

```python
#C. TREATING OUTLIERS USING CLIPPING
for i in num:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  df[i]=df[i].clip(lower,upper)
```

```python
#D. BOXPLOT AFTER OUTLIERS TREATMENT
plt.figure(figsize=(15,10))
df.select_dtypes(exclude=object).boxplot()
plt.xticks(rotation=45)
plt.title("Boxplot after outliers treatment")
plt.show()
```

```python
#NON-NUMERICAL COLUMN
obj=['Education', 'Marital_Status']
for i in obj:
  print(df[i].value_counts())
```

```python
#removing rows with unusual value
df = df[~df['Marital_Status'].isin(['YOLO', 'Absurd','Alone'])]
df['Marital_Status'].value_counts()
```

```python
#ENCODING USING ONE-HOT

df=pd.get_dummies(df,columns=['Education', 'Marital_Status'],drop_first=True)
df
```

```python
#input selection
x=df
x
```

```python
#FEATURE SCALING
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_scaled=scaler.fit_transform(x)
x_scaled
```

```python
#FINDING INERTIA FOR EACH K VALUES
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters=i,init="k-means++",random_state=41)
  kmeans.fit(x_scaled)
  wcss.append(kmeans.inertia_)
wcss
```

```python
#ELBOW METHOD
plt.plot(range(1,11),wcss)
plt.xlabel("K-Value")
plt.ylabel("Inertia")
plt.show()
```

```python
#MODEL BUILDING WHERE K=8
kmeans=KMeans(n_clusters=8,init="k-means++",random_state=41)
clust=kmeans.fit(x_scaled)
labels=kmeans.labels_
df['cluster']=kmeans.labels_
df
```

```python
df['cluster'].value_counts()
```

```python
#PERFORMANCE EVALUATION
from sklearn.metrics import silhouette_samples
print(silhouette_samples(x_scaled,labels))
```
