```python
#DATA CLEANING

#dropping unwanted column for analysis
df=df.drop('Address',axis=1)
```

```python
#duplicate value
print(df.duplicated().sum())
df=df.drop_duplicates()
print(df.duplicated().sum())
```

```python
#null values in target column
print(df['Price'].isnull().sum())
df=df.dropna(subset=['Price'])
print(df['Price'].isnull().sum())
```

```python
#checking min value of numerical column
num=df.drop('Price',axis=1).select_dtypes(exclude=object).columns
for i in num:
  print("min of",i,":",df[i].min())
```

```python
#replacing unusual zero to nan
zero_values=['Distance','Bedroom2','Bathroom','Landsize','BuildingArea']
for i in zero_values:
  df[i]=df[i].replace(0.0,np.nan)
```

```python
#rechecking the minimum value
for i in num:
  print("min of",i,":",df[i].min())
```

```python
#from date column extracting year and month
df['Date']=pd.to_datetime(df['Date'],format='mixed')
df['Year']=df['Date'].dt.year
df['Month']=df['Date'].dt.month
df=df.drop('Date',axis=1)
```

```python
#value count of month
df['Month'].value_counts()
```

```python
#null value handling
print(df.isnull().sum())
obj=df.select_dtypes(include=object).columns
num=df.drop('Price',axis=1).select_dtypes(exclude=object).columns
for i in obj:
  df[i]=df[i].fillna(df[i].mode()[0])
for i in num:
  df[i]=df[i].fillna(df[i].median())
```

```python
print(df.isnull().sum())
```

```python
#datatype
print(df.dtypes)
```

```python
list1=['Bedroom2','Bathroom','Car','YearBuilt','Propertycount']
for i in list1:
  df[i]=df[i].astype(int)
print(df.dtypes)
```

```python
#OUTLIERS

#visualization of outliers
import matplotlib.pyplot as plt
plt.figure(figsize=(15,10))
df.drop(['Price','Rooms','Month','Year'],axis=1).boxplot()
plt.xticks(rotation=45)
plt.title("Boxplot before outliers treatment")
plt.show()
```

```python
#checking number of outliers
out_num=df.drop(['Price','Rooms','Month','Year'],axis=1).select_dtypes(exclude=object).columns
for i in out_num:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outliers=df[(df[i]<lower)|(df[i]>upper)]
  print(i,"outliers:",len(outliers))
```

```python
#treating outliers
for i in out_num:
  Q1=df[i].quantile(0.25)
  Q3=df[i].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  df[i]=df[i].clip(lower,upper)
```

```python
#4.visualization  after treating outliers
plt.figure(figsize=(15,10))
df.drop(['Price','Rooms','Month','Year'],axis=1).boxplot()
plt.xticks(rotation=45)
plt.title("Boxplot after outliers treatment")
plt.show()
```

```python
#ENCODING
#ONE-HOT Encoding
# Check categorical columns
print(df.select_dtypes(include='object').columns)

df = pd.get_dummies(df,columns=['Type', 'Method', 'Regionname', 'CouncilArea'],drop_first=True)
print(df.head(3))
print(df.dtypes)
```

```python
#FREQUENCY ENCODING

# Frequency encoding for Suburb
suburb_freq = df['Suburb'].value_counts(normalize=True)
df['Suburb'] = df['Suburb'].map(suburb_freq)
# Frequency encoding for SellerG
seller_freq = df['SellerG'].value_counts(normalize=True)
df['SellerG'] = df['SellerG'].map(seller_freq)
```

```python
#FEATURE SELECTION
corr=df.corr(numeric_only=True)
print(corr['Price'].sort_values(ascending=False))
```

```python
#INPUT OUTPUT SEPARATION
x=df.drop(['Price','Bedroom2','BuildingArea','YearBuilt'],axis=1)
y=df['Price']
print(x.ndim)
print(y.ndim)
```

```python
#TRAIN-TEST-SPLITING
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.30,random_state=41)
print(xtrain.shape)
print(ytrain.shape)
print(xtest.shape)
print(ytest.shape)
```

```python
#HYPERPARAMETER TUNING OF LINEAR REGRESSION
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LinearRegression
lr_model=LinearRegression()
param={'fit_intercept':[True,False],'positive':[True,False]}
grid_lr=GridSearchCV(lr_model,param,cv=5,scoring='r2')
#training
grid_lr.fit(xtrain,ytrain)
#best performing
grid_lr.best_params_
```

```python
#LINEAR REGRESSION MODEL BUILDING
lr=LinearRegression(fit_intercept=True,positive= False)
#training
lr.fit(xtrain,ytrain)
#predicting
lr_ypred=lr.predict(xtest)
lr_ypred
```

```python
#REGRESSION PLOT
plt.scatter(ytest,lr_ypred)
plt.xlabel('actual price')
plt.ylabel('predicted price')
plt.title("actual vs predicted")
plt.show()
```

```python
#HYPERPARAMETER TUNING OF RANDOM FOREST
from sklearn.ensemble import RandomForestRegressor
rfr_model=RandomForestRegressor(random_state=41)
param={'n_estimators':[30,50],'max_depth':[5,10],'min_samples_split':[2,5],'min_samples_leaf':[1,2]}
grid_rfr=GridSearchCV(rfr_model,param,cv=3,scoring='r2',n_jobs=-1)
#training
grid_rfr.fit(xtrain,ytrain)
#best performing
grid_rfr.best_params_
```

```python
#RANDOM FOREST REGRESSION MODEL BUILDING
rfr=RandomForestRegressor(n_estimators=50,max_depth=10,min_samples_split=2,min_samples_leaf=2)
#training
rfr.fit(xtrain,ytrain)
#predicting
rfr_ypred=rfr.predict(xtest)
rfr_ypred
```

```python
#HYPERPARAMETER TUNING OF XGBOOST
from xgboost import XGBRegressor
xgb_model=XGBRegressor(random_state=41)
param={'n_estimators':[100,200],'max_depth':[3,5,7,10],'learning_rate':[0.01,0.1],'subsample':[0.8,1.0]}
grid_xgb=GridSearchCV(xgb_model,param,cv=5,scoring='r2')
#training
grid_xgb.fit(xtrain,ytrain)
#best performing
grid_xgb.best_params_
```

```python
#XGBOOST REGRESSION MODEL BUILDING
xgb=XGBRegressor(learning_rate=0.1,max_depth= 7, n_estimators=200, subsample=0.8)
#training
xgb.fit(xtrain,ytrain)
#predicting
xgb_ypred=xgb.predict(xtest)
xgb_ypred
```

```python
#HYPERPARAMETER TUNING OF DECISION TREE
from sklearn.tree import DecisionTreeRegressor
dt_model=DecisionTreeRegressor(random_state=41)
param = {'criterion': ['squared_error', 'friedman_mse'],'max_depth': [5, 10, 15],
         'min_samples_split': [2, 5],'min_samples_leaf': [1, 2]}

grid_dt=GridSearchCV(dt_model,param,cv=3,scoring='r2',n_jobs=-1)
#training
grid_dt.fit(xtrain,ytrain)
#best performing
grid_dt.best_params_
```

```python
#DECISION TREE MODEL BUILDING
dt=DecisionTreeRegressor(criterion='friedman_mse',max_depth=10,min_samples_leaf= 2,min_samples_split= 5)
#training
dt.fit(xtrain,ytrain)
#predicting
dt_ypred=dt.predict(xtest)
dt_ypred
```

```python
#TREE VISUALIZATION
from sklearn import tree
plt.figure(figsize=(10,8))
tree.plot_tree(dt,feature_names=x.columns,filled=True)
plt.show()
```

```python
#FEATURE SCALING
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(xtrain)
xtrain_scaled=scaler.transform(xtrain)
print(xtrain_scaled)
```

```python
xtest_scaled=scaler.transform(xtest)
xtest_scaled
```

```python
#HYPERPARAMETER TUNING FOR KNN
from sklearn.neighbors import KNeighborsRegressor
knn_model=KNeighborsRegressor()
param={'n_neighbors':[3,5,7,9,11],'weights':['uniform','distance'],'metric':['euclidean','manhattan']}
grid_knn=GridSearchCV(knn_model,param,cv=5,scoring='r2')
#training
grid_knn.fit(xtrain_scaled,ytrain)
#best performing
grid_knn.best_params_
```

```python
#K-NEAREST NEIGHBORS MODEL BUILDING
knn=KNeighborsRegressor(metric='manhattan', n_neighbors= 11, weights= 'distance')
#training
knn.fit(xtrain_scaled,ytrain)
#predicting
knn_ypred=knn.predict(xtest_scaled)
knn_ypred
```

```python
#TRAIN TEST SCORING
print("LR  Training Score:",lr.score(xtrain,ytrain),  "Testing score:",lr.score(xtest,ytest))
print("RFR Training Score:",rfr.score(xtrain,ytrain), "Testing score:",rfr.score(xtest,ytest))
print("XGB Training Score:",xgb.score(xtrain,ytrain), "Testing score:",xgb.score(xtest,ytest))
print("DT  Training Score:",dt.score(xtrain,ytrain),  "Testing score:",dt.score(xtest,ytest))
print("KNN Training Score:",knn.score(xtrain_scaled,ytrain), "Testing score:",knn.score(xtest_scaled,ytest))
```

```python
#PERFORMANCE EVALUATION
from sklearn.metrics import mean_squared_error,root_mean_squared_error,mean_absolute_error,r2_score
```

```python
#MEAN SQUARED ERROR
print("LR :",mean_squared_error(ytest,lr_ypred))
print("RFR :",mean_squared_error(ytest,rfr_ypred))
print("XGB :",mean_squared_error(ytest,xgb_ypred))
print("DT :",mean_squared_error(ytest,dt_ypred))
print("KNN :",mean_squared_error(ytest,knn_ypred))
```

```python
#ROOT MEAN SQUARED ERROR
print("LR :",root_mean_squared_error(ytest,lr_ypred))
print("RFR :",root_mean_squared_error(ytest,rfr_ypred))
print("XGB :",root_mean_squared_error(ytest,xgb_ypred))
print("DT :",root_mean_squared_error(ytest,dt_ypred))
print("KNN :",root_mean_squared_error(ytest,knn_ypred))
```

```python
#MEAN ABSOLUTE ERROR
print("LR :",mean_absolute_error(ytest,lr_ypred))
print("RFR :",mean_absolute_error(ytest,rfr_ypred))
print("XGB :",mean_absolute_error(ytest,xgb_ypred))
print("DT :",mean_absolute_error(ytest,dt_ypred))
print("KNN :",mean_absolute_error(ytest,knn_ypred))
```

```python
#R2 SCORE
print("LR :",r2_score(ytest,lr_ypred))
print("RFR :",r2_score(ytest,rfr_ypred))
print("XGB :",r2_score(ytest,xgb_ypred))
print("DT :",r2_score(ytest,dt_ypred))
print("KNN :",r2_score(ytest,knn_ypred))
```
