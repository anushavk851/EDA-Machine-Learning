```python
#DATA CLEANING

#duplicates
print("Train duplicates:", train.duplicated().sum())
print("Test duplicates:", test.duplicated().sum())
```

```python
#handling unusual values
for col in train.select_dtypes(include='object').columns:
    train[col] = train[col].replace(['–', 'not supported'], np.nan)

for col in test.select_dtypes(include='object').columns:
    test[col] = test[col].replace(['–', 'not supported'], np.nan)
```

```python
#Dropping unwanted columns
train.drop(['id', 'clean_title'], axis=1, inplace=True)
test.drop(['id', 'clean_title'], axis=1, inplace=True)
```

```python
#null values
print(train.isnull().sum())
print(test.isnull().sum())
```

```python
#null value handling in training dataset
train_obj=train.select_dtypes(include=object).columns
train_num=train.select_dtypes(exclude=object).columns
for ch in train_obj:
  train[ch]=train[ch].fillna(train[ch].mode()[0])
for i in train_num:
  train[i]=train[i].fillna(train[i].median())
```

```python
#null value handling in testing dataset
test_obj=test.select_dtypes(include=object).columns
test_num=test.select_dtypes(exclude=object).columns
for ch in test_obj:
  test[ch]=test[ch].fillna(train[ch].mode()[0])
for i in test_num:
  test[i]=test[i].fillna(test[i].median())
```

```python
#LABEL ENCODING FOR MODEL,ENGINE COLUMN
list1=['model','engine']
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
for i in list1:
  train[i]=le.fit_transform(train[i])
  test[i]=le.fit_transform(test[i])
```

```python
#ONE-HOT ENCODING FOR REMAINING COLUMNS
train = pd.get_dummies(train, drop_first=True)
test = pd.get_dummies(test, drop_first=True)
```

```python
train
```

```python
test
```

```python
#FEATURE AND TARGET SEPARATIONS
# Separate target
ytrain= train['price']
xtrain = train.drop('price',axis=1)
xtest = test
print(ytrain.shape)
print(xtrain.shape)
print(xtest.shape)
```

```python
# Align train and test columns
xtrain, xtest = xtrain.align(xtest,join='left',axis=1,fill_value=0)
```

```python
#FEATURE SCALING
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(xtrain)
xtrain=scaler.transform(xtrain)
xtrain
```

```python
xtest=scaler.transform(xtest)
xtest
```

```python
from sklearn.model_selection import GridSearchCV
#HYPERPARAMETER TUNING FOR Linear Regression
from sklearn.linear_model import LinearRegression
lr_model=LinearRegression()
lr_param={'fit_intercept':[True,False],'positive':[True,False]}
lr_grid=GridSearchCV(lr_model,lr_param,cv=3,scoring='r2')
lr_grid.fit(xtrain,ytrain)
lr_grid.best_params_
```

```python
#LR MODEL BUILDING
lr=LinearRegression(fit_intercept=True, positive= False)
lr.fit(xtrain,ytrain)
lr_ypred=lr.predict(xtest)
lr_ypred
```

```python
#HYPERPARAMETER TUNING FOR KNN
from sklearn.neighbors import KNeighborsRegressor
knn_model=KNeighborsRegressor()
knn_param={'n_neighbors':[5,7,9,11],'weights':['distance','random'],'metric':['euclidean','manhattan']}
knn_grid=GridSearchCV(knn_model,knn_param,cv=3,scoring='r2')
knn_grid.fit(xtrain,ytrain)
knn_grid.best_params_
```

```python
#KNN MODEL BUILDING
knn=KNeighborsRegressor(metric= 'euclidean', n_neighbors=5, weights='distance')
knn.fit(xtrain,ytrain)
knn_ypred=knn.predict(xtest)
knn_ypred
```

```python
#HYPERPARAMETER TUNING FOR DT
from sklearn.tree import DecisionTreeRegressor
dt_model=DecisionTreeRegressor(random_state=41)
dt_param={'criterion':['squared_error','friedman_mse','absolute_error'],'max_depth':[5,7,10,None],
          'min_samples_split':[2,5,10],'min_samples_leaf':[1,2,4]}
dt_grid=GridSearchCV(dt_model,dt_param,cv=3,scoring='r2')
dt_grid.fit(xtrain,ytrain)
dt_grid.best_params_
```

```python
#DECISION TREE REGRESSION MODEL BUILDING
dt=DecisionTreeRegressor(criterion= 'squared_error',max_depth= 7,min_samples_leaf= 1,min_samples_split=2,random_state=41)
dt.fit(xtrain,ytrain)
dt_ypred=dt.predict(xtest)
dt_ypred
```

```python
#HYPERPARAMETER TUNING OF RANDOM FOREST
from sklearn.ensemble import RandomForestRegressor
rfr_model=RandomForestRegressor(random_state=41)
rfr_param={'n_estimators':[30,50],'max_depth':[5,10],'min_samples_split':[2,5],'min_samples_leaf':[1,2]}
rfr_grid=GridSearchCV(rfr_model,rfr_param,cv=3,scoring='r2',)
rfr_grid.fit(xtrain,ytrain)
rfr_grid.best_params_
```

```python
#RANDOM FOREST REGRESSION MODEL BUILDING
rfr=RandomForestRegressor(n_estimators=50,max_depth=10,min_samples_split=2,min_samples_leaf=2)
rfr.fit(xtrain,ytrain)
rfr_ypred=rfr.predict(xtest)
rfr_ypred
```

```python
#HYPERPARAMETER TUNING OF XGBOOST
from xgboost import XGBRegressor
xgb_model=XGBRegressor(random_state=41)
xgb_param={'n_estimators':[100,200],'max_depth':[3,5,7,10],'learning_rate':[0.01,0.1],'subsample':[0.8,1.0]}
xgb_grid=GridSearchCV(xgb_model,xgb_param,cv=5,scoring='r2')
xgb_grid.fit(xtrain,ytrain)
xgb_grid.best_params_
```

```python
#XGBOOST REGRESSION MODEL BUILDING
xgb=XGBRegressor(learning_rate=0.1,max_depth= 7, n_estimators=200, subsample=0.8)
xgb.fit(xtrain,ytrain)
xgb_ypred=xgb.predict(xtest)
xgb_ypred
```

```python
#DIVIDING THE TRAIN DATASET TO TRAINING AND TESTING DATASET FOR EVALUATING THE PERFORMANCE OF MODEL
x1=train.drop('price',axis=1)
y1=train['price']
x1
```

```python
#TRAIN-TEST-SPLITTING
from sklearn.model_selection import train_test_split
x1train,x1test,y1train,y1test=train_test_split(x1,y1,test_size=0.30,random_state=41)
print(x1train.shape)
print(x1test.shape)
print(y1train.shape)
print(y1test.shape)
```

```python
#FEATURE SCALING
scale=StandardScaler()
scale.fit(x1train)
x1train=scale.transform(x1train)
x1train
```

```python
x1test=scale.transform(x1test)
x1test
```

```python
#LR MODEL BUILDING
lr1=LinearRegression(fit_intercept=True, positive= False)
lr1.fit(x1train,y1train)
lr1_ypred=lr1.predict(x1test)
lr1_ypred
```

```python
#KNN MODEL BUILDING
knn1=KNeighborsRegressor(metric= 'euclidean', n_neighbors=5, weights='distance')
knn1.fit(x1train,y1train)
knn1_ypred=knn1.predict(x1test)
knn1_ypred
```

```python
#DECISION TREE REGRESSION MODEL BUILDING
dt1=DecisionTreeRegressor(criterion= 'squared_error',max_depth= 7,min_samples_leaf= 1,min_samples_split=2,random_state=41)
dt1.fit(x1train,y1train)
dt1_ypred=dt1.predict(x1test)
dt1_ypred
```

```python
#RANDOM FOREST REGRESSION MODEL BUILDING
rfr1=RandomForestRegressor(n_estimators=50,max_depth=10,min_samples_split=2,min_samples_leaf=2)
rfr1.fit(x1train,y1train)
rfr1_ypred=rfr1.predict(x1test)
rfr1_ypred
```

```python
#XGBOOST REGRESSION MODEL BUILDING
xgb1=XGBRegressor(learning_rate=0.1,max_depth= 7, n_estimators=200, subsample=0.8)
xgb1.fit(x1train,y1train)
xgb1_ypred=xgb1.predict(x1test)
xgb1_ypred
```

```python
#TRAIN TEST SCORING
print("LR  Training Score:",lr1.score(x1train,y1train),  "Testing score:",lr1.score(x1test,y1test))
print("RFR Training Score:",rfr1.score(x1train,y1train), "Testing score:",rfr1.score(x1test,y1test))
print("XGB Training Score:",xgb1.score(x1train,y1train), "Testing score:",xgb1.score(x1test,y1test))
print("DT  Training Score:",dt1.score(x1train,y1train),  "Testing score:",dt1.score(x1test,y1test))
print("KNN Training Score:",knn1.score(x1train,y1train), "Testing score:",knn1.score(x1test,y1test))
```

```python
#PERFORMANCE EVALUATION
from sklearn.metrics import mean_squared_error,root_mean_squared_error,mean_absolute_error,r2_score
```

```python
#MEAN SQUARED ERROR
print("LR :",mean_squared_error(y1test,lr1_ypred))
print("RFR :",mean_squared_error(y1test,rfr1_ypred))
print("XGB :",mean_squared_error(y1test,xgb1_ypred))
print("DT :",mean_squared_error(y1test,dt1_ypred))
print("KNN :",mean_squared_error(y1test,knn1_ypred))
```

```python
#ROOT MEAN SQUARED ERROR
print("LR :",root_mean_squared_error(y1test,lr1_ypred))
print("RFR :",root_mean_squared_error(y1test,rfr1_ypred))
print("XGB :",root_mean_squared_error(y1test,xgb1_ypred))
print("DT :",root_mean_squared_error(y1test,dt1_ypred))
print("KNN :",root_mean_squared_error(y1test,knn1_ypred))
```

```python
#MEAN ABSOLUTE ERROR
print("LR :",mean_absolute_error(y1test,lr1_ypred))
print("RFR :",mean_absolute_error(y1test,rfr1_ypred))
print("XGB :",mean_absolute_error(y1test,xgb1_ypred))
print("DT :",mean_absolute_error(y1test,dt1_ypred))
print("KNN :",mean_absolute_error(y1test,knn1_ypred))
```

```python
#R2 SCORE
print("LR :",r2_score(y1test,lr1_ypred))
print("RFR :",r2_score(y1test,rfr1_ypred))
print("XGB :",r2_score(y1test,xgb1_ypred))
print("DT :",r2_score(y1test,dt1_ypred))
print("KNN :",r2_score(y1test,knn1_ypred))
```
